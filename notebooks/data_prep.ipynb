```python
import numpy as np
import pandas as pd
pd.set_option('display.max_info_columns', 1000)
pd.set_option('display.max_columns', None) 
from pathlib import Path
```

```python
data_path = Path.cwd().parent / 'Data'
business_groups_path = data_path / 'VK_CaseChamp_Dataset.xlsx'

Business_groups = pd.read_excel(business_groups_path, sheet_name='Бизнесы')
Market_Partner = pd.read_excel(business_groups_path, sheet_name='Маркет Вк и Партнер сводка')
Market_Total = pd.read_excel(business_groups_path, sheet_name='Итоги Маркет Вк')
Market_Partner_dynamics = pd.read_excel(business_groups_path, sheet_name='Маркет Вконтакте — по дням')
Market_Partner_traffic = pd.read_excel(business_groups_path, sheet_name='Маркет Вконтакте — источники')
Market_Partner_categories = pd.read_excel(business_groups_path, sheet_name='Маркет Вконтакте — категории')
Market_Partner_gender_age = pd.read_excel(business_groups_path, sheet_name='Покупатели')
Market_Partner_clients_by_regions = pd.read_excel(business_groups_path, sheet_name='Покупатели по регионам')
```

• Business_groups **«Бизнесы»** — данные о бизнес-сообществах ВКонтакте: тематика и тип бизнеса, активность и
вовлечённость за месяц, подключённые инструменты торговли (каталог, чекаут, доставка, оплата), заказы и траты на
рекламу.

• Market_Partner **«Маркет ВКонтакте и Партнер»** — сводные метрики Маркета ВКонтакте и Партнёра за месяц: показы,
открытия карточки, заказы, GMV, конверсии, покупатели.

• Market_Total **«Итоги Маркет Вк»** — общие метрики Маркета ВКонтакте без деления на
сегменты.

• Market_Partner_dynamics **«Маркет ВКонтакте и Партнера — по дням»** — ежедневная динамика ключевых метрик Маркета Вконтакте и Партнёра
за месяц 2026-09-01 — 2026-09-30.

• Market_Partner_traffic **«Маркет ВКонтакте и Партнер — источники»** — распределение открытий карточки товара по источникам трафика для
Маркета ВКонтакте и Партнёра.

• Market_Partner_categories **«Маркет Вконтакте И и Партнер — категории»** — открытия карточки и заказы по категориям товаров для Маркета
ВКонтакте и Партнёра.

• Market_Partner_gender_age **«Покупатели по полу и возрасту»** — число покупателей и заказов по полу и возрасту для Маркета ВКонтакте и Партнёра.

• Market_Partner_clients_by_regions **«Покупатели по регионам»** — число покупателей и заказов по регионам России для Маркета ВКонтакте и
Партнёра
